# Count Grid Points and Zone Cycles

## Carpet

* physical extent: [-400,-400,-400] : [400,400,400]
* interior extent: [-396.667,-396.667,-396.667] : [396.667,396.667,396.667]
* exterior extent: [-406.667,-406.667,-406.667] : [406.667,406.667,406.667]

```julia
dx0 = 10/3
```

```julia
ExteriorPointGridCarpet = [
    [[245,245,245]],
    [[287,283,283]],
    [[293,283,283]],
    [[303,283,283]],
    [[297,259,259]],
    [[287,211,211]],
    [[365,211,211]],
    [[172,173,173],[172,173,173]],
    [[173,173,173],[173,173,173]],
]
```

```julia
for d in ExteriorPointGridCarpet
    s = sum([prod([x for x in xs]) for xs in d])
    println(s)
end
```

```julia
ExteriorCellGridCarpet = [[xs .- 1 for xs in d] for d in ExteriorPointGridCarpet]
```

```julia
InteriorCellGridCarpet = []
for (i, d) in enumerate(ExteriorCellGridCarpet)
    if i == 1
        push!(InteriorCellGridCarpet, [xs .- 6 for xs in d])
    else
        push!(InteriorCellGridCarpet, [xs .- 18 for xs in d])
    end
end
InteriorCellGridCarpet
```

### Ratio Between Subcycling and Non-Subcycling

```julia
CyclesPerFullIterationNonSubcycling = 0
for d in InteriorCellGridCarpet
    factor = 2^8
    for xs in d
        CyclesPerFullIterationNonSubcycling += prod(xs) * factor
    end
end
println(CyclesPerFullIterationNonSubcycling)
```

```julia
CyclesPerFullIterationSubcycling = 0
for (l, d) in enumerate(InteriorCellGridCarpet)
    factor = 2^(l-1)
    for xs in d
        CyclesPerFullIterationSubcycling += prod(xs) * factor
    end
end
println(CyclesPerFullIterationSubcycling)
```

```julia
println(CyclesPerFullIterationNonSubcycling / CyclesPerFullIterationSubcycling)
```
