## Carpet

* physical extent: [-400,-400,-400] : [400,400,400]
* interior extent: [-396.667,-396.667,-396.667] : [396.667,396.667,396.667]
* exterior extent: [-406.667,-406.667,-406.667] : [406.667,406.667,406.667]

```julia
dx0 = 10/3
```

```julia
ExteriorGridCarpet = [
    [[245,245,245]],
    [[287,283,283]],
    [[293,283,283]],
    [[303,283,283]],
    [[297,259,259]],
    [[287,211,211]],
    [[365,211,211]],
    [[172,173,173],[172,173,173]],
    [[173,173,173],[173,173,173]],
]
```

#### Level 1

```julia
(406.667 - 396.667) / dx0  # There are 3 boundary points
```

```julia
ExteriorGridCarpet[1][1] .- 6  # interior points
```
